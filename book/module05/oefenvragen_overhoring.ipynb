### Vraag 1 — Doel A1 (K)

Leg in eigen woorden uit waarom een computer een geluidsgolf niet exact kan voorstellen, maar enkel een benadering. Gebruik daarbij de woorden **continu**, **discreet**, **sample** en **samplerate**.

```{admonition} Oplossing
:class: dropdown

In de fysica is een geluidsgolf een **continu** signaal: op elk moment in de tijd bestaat er een drukwaarde. Een computer kan echter niet met een oneindig aantal tijdstippen werken. Daarom maken we het signaal **discreet**: we meten de waarde op een beperkt aantal momenten. Elk zo'n meetpunt heet een **sample**. Het aantal samples per seconde noemen we de **samplerate**. Hoe hoger de samplerate, hoe nauwkeuriger de benadering.
```

### Vraag 5 — Doel A3 (K)

In de formule $y = A \sin(2\pi f t)$:

(a) Wat doet de parameter $A$?  
(b) Wat doet de parameter $f$?  
(c) Waarom staat er $2\pi$ in de formule? Wat zou er gebeuren als je die factor zou weglaten?

```{admonition} Oplossing
:class: dropdown

(a) $A$ is de **amplitude**: de maximale uitwijking van de golf. Bij geluid bepaalt dit het volume.

(c) De sinusfunctie werkt met **radialen**. Eén volledige golf komt overeen met een hoek van $2\pi$ radialen. Als je $2\pi$ zou weglaten, zou één seconde overeenkomen met slechts ongeveer $1/(2\pi) \approx 0.16$ golf — geen volledige cyclus.
```

### Vraag 10 — Doel A5 (S)

````python
import numpy as np
import matplotlib.pyplot as plt

SAMPLERATE = 1000
duration = 2.0
frequency = 10

t = np.linspace(0, duration, int(SAMPLERATE * duration), endpoint=False)
# lineaire demping: van 1 naar 0 over de hele duur
demping = 1 - t / duration
wave = np.sin(2 * np.pi * frequency * t) * demping

plt.plot(t, wave)
plt.xlabel("Tijd (s)")
plt.ylabel("Amplitude")
plt.title("Lineair gedempte sinus 10 Hz")
plt.show()
````

## Module B — Klankkleur

### Vraag 14 — Doel B4 (K)

Wat is een **boventoon**, en wat is een **grondtoon**? Als de grondtoon 150 Hz is, wat zijn dan de eerste vier boventonen? (Let op: tellen we de grondtoon zelf mee als eerste boventoon? Verklaar je antwoord.)

```{admonition} Oplossing
:class: dropdown

De **grondtoon** $f_0$ is de basisfrequentie van een geluid. Een **boventoon** is een sinus waarvan de frequentie een geheel veelvoud is van de grondtoon: $n \cdot f_0$ voor $n = 1, 2, 3, \ldots$

In de cursus wordt $n = 1$ (de grondtoon zelf) mee opgenomen in de reeks. De eerste vier "boventonen" in die nummering zijn dus:

* $n=1$: 150 Hz (de grondtoon)
* $n=2$: 300 Hz
* $n=3$: 450 Hz
* $n=4$: 600 Hz
```

### Vraag 15 — Doel B4 (S)

Schrijf een functie `make_harmonic(fundamental, n)` die de $n$-de boventoon van een grondtoon teruggeeft als numpy-array. Gebruik samplerate 16000 en duur 1 seconde. Test je functie door drie aanroepen te doen: voor $n = 1, 2, 3$ bij grondtoon 220 Hz, en print telkens de eerste vijf waarden.

````python
import numpy as np

SAMPLERATE = 16000
duration = 1.0
t = np.linspace(0, duration, int(SAMPLERATE * duration), endpoint=False)

def make_harmonic(fundamental, n):
    return np.sin(2 * np.pi * n * fundamental * t)

for n in [1, 2, 3]:
    harmonic = make_harmonic(220, n)
    print(f"n={n}: {harmonic}")
````

### Vraag 16 — Doel B5 (K)

Twee instrumenten spelen exact dezelfde noot (bv. een A van 440 Hz), en toch klinken ze duidelijk anders. Hoe komt dat? Leg uit aan de hand van het begrip **klankkleur**.

```{admonition} Oplossing
:class: dropdown

Een echte instrumenttoon bestaat niet uit één enkele sinus, maar uit de **grondtoon plus een reeks boventonen** (sinussen met frequentie $n \cdot f_0$). Elk instrument heeft een eigen "vingerafdruk" in welke boventonen sterk klinken en welke zwak. Het verschil zit dus in de **amplitudeverhoudingen** van de boventonen. Dat verschil in karakter noemen we de **klankkleur** of het **timbre**. De grondtoon (en dus de toonhoogte) is gelijk, maar de mix van boventonen verschilt.
```

## Module C — Envelopes

### Vraag 21 — Doel C1 (K)

Wat is een **envelope** in de context van geluid? Welke waarden bevat ze, en welke bewerking gebruik je om ze op een golf toe te passen? Wat gebeurt er als de envelope en de golf niet even lang zijn?

```{admonition} Oplossing
:class: dropdown

Een **envelope** is een array van amplitudes (waarden tussen 0 en 1) — één per sample — die beschrijft hoe het volume van een geluid in de tijd verandert. Je past haar toe op een golf via **elementsgewijze vermenigvuldiging**: `wave * envelope`.

Als de envelope en de golf niet even lang zijn, geeft numpy een **fout** (ValueError) omdat de vormen niet bij elkaar passen. De twee arrays moeten dezelfde lengte hebben.


```

### Vraag 22 — Doel C1 (S)

Schrijf code die een sinus van 440 Hz over 1 seconde genereert (samplerate 16000), en die vermenigvuldigt met een envelope die over de hele duur **lineair stijgt van 0 naar 1**. Plot de golf, de envelope en het product op één figuur.

````python
import numpy as np
import matplotlib.pyplot as plt

SAMPLERATE = 16000
duration = 1.0
t = np.linspace(0, duration, int(SAMPLERATE * duration), endpoint=False)

wave = np.sin(2 * np.pi * 440 * t)
envelope = t / duration  # lineair van 0 naar 1
product = wave * envelope

plt.figure(figsize=(10, 4))
plt.plot(t, wave, alpha=0.3, label="golf")
plt.plot(t, envelope, color="red", linewidth=2, label="envelope")
plt.plot(t, product, color="black", label="product")
plt.legend()
plt.xlabel("Tijd (s)")
plt.show()
````

### Vraag 24 — Doel C5 (T)

Bekijk volgende code voor een AD-envelope:

```python
def make_ad_envelope(duration, attack, decay):
    n_samples = int(SAMPLERATE * duration)
    attack_samples = int(SAMPLERATE * attack)
    decay_samples = int(SAMPLERATE * decay)
    
    envelope = np.zeros(n_samples)
    for n in range(n_samples):
        if n < attack_samples:
            envelope[n] = n / attack_samples
        elif n < attack_samples + decay_samples:
            envelope[n] = 1.0 - (n - attack_samples) / decay_samples
    return envelope
```

Stel `SAMPLERATE = 16000` en je roept aan: `make_ad_envelope(duration=2.0, attack=0.2, decay=1.0)`.

(a) Bij welke sample-index `n` schakelt de envelope over van "stijgend" naar "dalend"?  
(b) Bij welke sample-index `n` is de envelope terug op 0 (einde van de decay)?  
(c) Wat staat er na die index in de envelope, en waarom?

```{admonition} Oplossing
:class: dropdown

Eerst de samples berekenen:

* `n_samples = 16000 · 2.0 = 32000`
* `attack_samples = 16000 · 0.2 = 3200`
* `decay_samples = 16000 · 1.0 = 16000`

(a) De overschakeling gebeurt bij `n = attack_samples = 3200`. Op dat moment bereikt de envelope haar maximum 1.0.

(b) De decay is klaar bij `n = attack_samples + decay_samples = 3200 + 16000 = 19200`. Daar bereikt de envelope 0.

(c) Na index 19200 staat overal 0. Dat komt omdat de array initieel met `np.zeros(n_samples)` werd aangemaakt, en de `if/elif`-voorwaarden geen `else`-tak hebben. Voor alle `n` na de decay-fase wordt de waarde dus niet aangepast — ze blijft 0.
```

### Vraag 25 — Doel C5 (S)

Schrijf vanaf nul een functie `make_ad_envelope(duration, attack, decay)` met behulp van een `for`-lus. De envelope:

* heeft in totaal `duration` seconden
* stijgt lineair van 0 naar 1 over `attack` seconden
* daalt daarna lineair van 1 naar 0 over `decay` seconden
* blijft op 0 voor de resterende tijd (als die er is)

Test je functie door ze toe te passen op een sinus van 220 Hz met duur 1 s, attack 0.1 s en decay 0.5 s. Plot het resultaat.

````python
import numpy as np
import matplotlib.pyplot as plt

SAMPLERATE = 16000

def make_ad_envelope(duration, attack, decay):
    n_samples = int(SAMPLERATE * duration)
    attack_samples = int(SAMPLERATE * attack)
    decay_samples = int(SAMPLERATE * decay)

    envelope = np.zeros(n_samples)
    for n in range(n_samples):
        if n < attack_samples:
            envelope[n] = n / attack_samples
        elif n < attack_samples + decay_samples:
            envelope[n] = 1.0 - (n - attack_samples) / decay_samples
    return envelope

duration = 1.0
t = np.linspace(0, duration, int(SAMPLERATE * duration), endpoint=False)
wave = np.sin(2 * np.pi * 220 * t)
envelope = make_ad_envelope(duration, attack=0.1, decay=0.5)

plt.plot(t, wave * envelope)
plt.xlabel("Tijd (s)")
plt.ylabel("Amplitude")
plt.title("Sinus 220 Hz met AD-envelope (attack 0.1 s, decay 0.5 s)")
plt.show()
````
